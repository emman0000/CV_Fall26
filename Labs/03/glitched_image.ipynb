```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("chip.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
```

```python
h, w = image.shape[:2]
```

```python
theta = 45
```

```python
sx = 1.5
sy = 1.5
S = np.float32([[sx,0,0],[0,sy,0],[0,0,1]])

kx = 0.2
ky = 0

R = np.float32([[np.cos(theta), -np.sin(theta), 0], [np.sin(theta), np.cos(theta), 0], [0,0,1]])
H = np.float32([[1,kx,0],[ky,1,0],[0,0,1]])

tx = 50
ty = 30

T = np.float32([[1,0,tx], [0,1,ty], [0,0,1]])

M = T @ R @ H @ S
```

```python
cx = w/2
cy = h/2
```

```python
tx = cx - cx*np.cos(theta) + cy*np.sin(theta)

ty = cy - cx*np.sin(theta) - cy*np.cos(theta)
```

```python
center = (cx,cy)
```

```python
M_affine = M[:2,:]
```

```python
r = cv2.warpAffine(image , M_affine , (w,h))
```

```python
plt.imshow(r)
plt.axis("off")
plt.show()
```
