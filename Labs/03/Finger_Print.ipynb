```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("fingerprint.jpg")
plt.imshow(image)
```

```python
h, w = image.shape[:2]
cx = w/2
cy = h/2
```

```python
sx = 3
sy = 3
```

```python
s = np.array([[sx, 0],[0,sy]])
```

```python
tx = (1 - sx) * cx
ty = (1 - sy) * cy
```

```python
M = np.float32([[sx, 0 , tx], [0, sy, ty]])
```

```python
transform = cv2.warpAffine(image, M, (w,h))
```

```python
plt.imshow(cv2.cvtColor(transform, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
