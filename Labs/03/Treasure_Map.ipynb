```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create a white 200 x 300 map
treasure_map = np.ones((200, 300), dtype=np.uint8) * 255

# Draw a simple map feature
cv.rectangle(treasure_map, (80, 60), (220, 140), 0, 2)

# Draw an X to represent the treasure
cv.line(treasure_map, (130, 80), (170, 120), 0, 3)
cv.line(treasure_map, (170, 80), (130, 120), 0, 3)

# Show the map
plt.imshow(treasure_map, cmap="gray")
plt.axis("off");
```

```python
# Move the map 150 pixels right and 80 pixels down
tx = 150
ty = 80

# 3x3 translation matrix
translation_matrix = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
])

print("Translation Matrix:")
print(translation_matrix)
```

```python
# Convert the translation into OpenCV's 2x3 format
M = np.array([
    [1, 0, translation_matrix[0, 2]],
    [0, 1, translation_matrix[1, 2]]
], dtype=np.float32)

# Move the map
translated_map = cv2.warpAffine(
    treasure_map,
    M,
    (treasure_map.shape[1], treasure_map.shape[0])
)

# Show the moved map
plt.imshow(translated_map, cmap="gray")
plt.axis("off");
```
