```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("chip.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
```

```python
h, w = image.shape[:2]
```

```python
cx = w/2
cy = h/2
```

```python
theta = 45
```

```python
tx = cx - cx*np.cos(theta) + cy*np.sin(theta)

ty = cy - cx*np.sin(theta) - cy*np.cos(theta)
```

```python
M = np.float32([[np.cos(theta), -np.sin(theta), tx], [np.sin(theta), np.cos(theta), ty], [0,0,1]])
```

```python
center = (cx,cy)
```

```python
M= cv2.getRotationMatrix2D(center,theta,1)
rotated = cv2.warpAffine(image, M, (w, h))
```

```python
plt.imshow(rotated)
plt.axis("off")
plt.show()
```
