```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
fingerprint = np.zeros((100,100), dtype=np.uint8)
fingerprint[40:60, 40:60] = 255
plt.imshow(fingerprint,cmap="gray")
plt.axis("off");
```

```python
# 300% means 3 times bigger
scale_matrix = np.array([
    [3, 0],
    [0, 3]
])

print(scale_matrix)
```

```python
# Original image size
height, width = fingerprint.shape

# Calculate how much to move the enlarged image
tx = -(3 * width - width) // 2
ty = -(3 * height - height) // 2

print("X translation:", tx)
print("Y translation:", ty)
```

```python
new_width = width * 3
new_height = height * 3

scaled = cv2.resize(fingerprint, (new_width, new_height))

#display scaled image
plt.imshow(scaled, cmap="gray")
plt.axis("off");
```

```python
centered= np.zeros((height, width),dtype=np.uint8)

start_x = (width - new_width) // 2
start_y = (height - new_height) // 2

centered = np.zeros((new_height, new_width), dtype=np.uint8)
x=(new_width - width)//2
y=(new_height - height)//2

centered[y:y+height, x:x+width] = fingerprint
plt.imshow(centered, cmap="gray")
plt.axis("off");
```
