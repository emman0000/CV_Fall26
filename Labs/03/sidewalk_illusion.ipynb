```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread("chalk3.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
```

```python
h, w = image.shape[:2]
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.grid()
plt.show()
```

```python
src = np.float32([
    [162, 230],   # top
    [313, 320],   # right
    [163, 483],   # bottom
    [8, 341]      # left
])
```

```python
dst = np.float32([
    [0, 0],       # top-left
    [500, 0],     # top-right
    [500, 500],   # bottom-right
    [0, 500]      # bottom-left
])
```

```python
M = cv2.getPerspectiveTransform(src,dst)
```

```python
result = cv2.warpPerspective(image, M , (500,500))
```

```python
plt.imshow(result)
plt.axis("off")
plt.show()
```
