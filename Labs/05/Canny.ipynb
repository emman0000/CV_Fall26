```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("cancan.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = img
print(img.shape)
plt.axis("off")
plt.imshow(img)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.axis("off")
plt.imshow(gray, cmap='gray')
```

```python
edges1 = cv2.Canny(gray , 50, 150)
```

```python
plt.imshow(edges1, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis('off')
plt.show()
```

```python
edges2 = cv2.Canny(gray, 50, 100)
edges3 = cv2.Canny(gray, 10, 20)
edges4 = cv2.Canny(gray, 200, 300)
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2,2,1)
plt.imshow(edges2, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(edges3, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(edges4, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")

plt.show()
```
