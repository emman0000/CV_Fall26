```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("car.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = img
print(img.shape)
plt.axis("off")
plt.imshow(img)
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
```

```python
lower_green = np.array([36, 25 ,25])
upper_green = np.array([86, 255, 255])
```

```python
mask = cv2.inRange(hsv,lower_green,upper_green)
```

```python
plt.imshow(mask, cmap='gray')
plt.title("Color Mask")
plt.axis('off')
plt.show()
```

```python
result = cv2.bitwise_and(img, img, mask=mask)
```

```python
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Extracted Object")
plt.axis('off')
plt.show()
```

```python
lower_wide = np.array([15, 70, 70])
upper_wide = np.array([45, 255, 255])
```

```python
mask_wide = cv2.inRange(
    hsv,
    lower_wide,
    upper_wide
)
```

```python
plt.imshow(mask_wide, cmap='gray')
plt.title("Color Mask")
plt.axis('off')
plt.show()
```

```python
result2 = cv2.bitwise_and(img, img, mask=mask_wide)
```

```python
plt.imshow(cv2.cvtColor(result2, cv2.COLOR_BGR2RGB))
plt.title("Extracted Object")
plt.axis('off')
plt.show()
```

Too narrow → object gets chopped up.
Too wide → background gets included.
