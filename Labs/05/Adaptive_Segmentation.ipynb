```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("doc.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(img.shape)
plt.axis("off")
plt.imshow(img)
```

```python
block_sizes = [11,21,31]
c_val = [2,5,10]
```

```python
def adaptive_mean(gray, block_size, c):
   return cv2.adaptiveThreshold(
       gray , 255,cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, block_size, c)
```

```python
def adaptive_gaussian(gray, block_size,c):
  return cv2.adaptiveThreshold(gray,255,cv2.ADAPTUVETH_GAUSSIAN_C, cv2.THRESH_BINARY, block_size, c)
```

```python
r = adaptive_mean(gray,11,5)
plt.imshow(r, cmap='gray')
plt.axis("off")
plt.title("Mean with Block size =11 and C = 5")
plt.show()
```

```python
r = adaptive_mean(gray,21,5)
plt.imshow(r, cmap='gray')
plt.axis("off")
plt.title("Mean with Block size =11 and C = 5")
plt.show()
```

```python
r = adaptive_mean(gray,31,5)
plt.imshow(r, cmap='gray')
plt.axis("off")
plt.title("Mean with Block size =11 and C = 5")
plt.show()
```

```python
result1 = adaptive_mean(gray, 21, 2)
result2 = adaptive_mean(gray, 21, 5)
result3 = adaptive_mean(gray, 21, 10)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(result1, cmap='gray')
plt.title("C = 2")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(result2, cmap='gray')
plt.title("C = 5")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(result3, cmap='gray')
plt.title("C = 10")
plt.axis('off')

plt.show()
```

When the neighbourhood is too small: The result can become noisy and sensitive to small intensity changes.

When the neighbourhood is too large: Local details may be lost and the threshold may not adapt well to small lighting changes.

When C is increased: The local threshold decreases, causing more pixels to be classified as white.

Cleanest text segmentation: The combination of blockSize = 21 and C = 5 produced the cleanest result.

Better adaptive method: Gaussian adaptive thresholding performed better because it produced cleaner text with fewer unwanted regions.
