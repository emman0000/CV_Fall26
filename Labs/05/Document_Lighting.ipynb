```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("doc.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
print(img.shape)
plt.axis("off")
plt.imshow(img)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(gray.shape)
```

```python
_, th1 = cv2.threshold(gray, 80,255, cv2.THRESH_BINARY)
_, th2 = cv2.threshold(gray, 120,255, cv2.THRESH_BINARY)
_, th3 = cv2.threshold(gray, 160,255, cv2.THRESH_BINARY)
```

```python
plt.figure(figsize=(12,10))

plt.subplot(2,2,1)
plt.imshow(gray, cmap='gray')
plt.title("OG Image")
plt.axis("off")


plt.subplot(2,2,2)
plt.imshow(th1, cmap='gray')
plt.title("Threshold 80")
plt.axis("off")


plt.subplot(2,2,3)
plt.imshow(th2, cmap='gray')
plt.title("Threshold 120")
plt.axis("off")


plt.subplot(2,2,4)
plt.imshow(th3, cmap='gray')
plt.title("Threshold 160")
plt.axis("off")

plt.show()
```

A single global threshold struggles when illumination varies because different regions of the image have different intensity ranges. A threshold suitable for a bright region may fail in a darker region, causing text or background pixels to be incorrectly classified.
