```python
#T1

class GroceryManager:

    def __init__(self):
        self.items = []

    def add_item(self, item, quantity, price):
        self.items.append({
            "item": item,
            "quantity": quantity,
            "price": price
        })

    def remove_item(self, item):
        for product in self.items:
            if product["item"] == item:
                self.items.remove(product)
                print(item, "removed.")
                return

        print("Error:", item, "does not exist.")

    def view_list(self):
        for product in self.items:
            print(
                product["item"],
                "| Quantity:", product["quantity"],
                "| Price:", product["price"]
            )

    def calculate_total(self):
        total = 0

        for product in self.items:
            total += product["quantity"] * product["price"]

        return total


# Test the class
grocery = GroceryManager()

grocery.add_item("Milk", 2, 200)
grocery.add_item("Bread", 1, 150)
grocery.add_item("Eggs", 12, 30)

print("Grocery List:")
grocery.view_list()

print("\nTotal cost:", grocery.calculate_total())

grocery.remove_item("Bread")

print("\nAfter removing Bread:")
grocery.view_list()
```

```python
#T2

students = {
    101: {
        "Name": "Ali",
        "Major": "Computer Science",
        "Grades": [80, 75, 90]
    },

    102: {
        "Name": "Sara",
        "Major": "Electrical Engineering",
        "Grades": [85, 95, 90]
    },

    103: {
        "Name": "Ahmed",
        "Major": "Computer Science",
        "Grades": [70, 80, 75]
    }
}


# Find student with highest average
def highest_average(students):

    best_name = ""
    best_average = 0

    for student in students.values():

        average = sum(student["Grades"]) / len(student["Grades"])

        if average > best_average:
            best_average = average
            best_name = student["Name"]

    return best_name


# Search students by major
def search_by_major(students, major):

    for student in students.values():

        if student["Major"] == major:
            print(student["Name"])


print("Student with highest average:")
print(highest_average(students))

print("\nComputer Science students:")
search_by_major(students, "Computer Science")
```

```python
#T5

image_path = "Sukuna.jpeg"

image = cv2.imread(image_path)

if image is None:
    print("Error: Image not found.")

else:
    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Apply heavy Gaussian blur
    blurred = cv2.GaussianBlur(
        image_rgb,
        (25, 25),
        0
    )

    # Get image size
    height, width = image_rgb.shape[:2]

    # Find center
    center_x = width // 2
    center_y = height // 2

    # We want a 300 x 300 ROI
    roi_size = 300

    x1 = center_x - roi_size // 2
    x2 = center_x + roi_size // 2

    y1 = center_y - roi_size // 2
    y2 = center_y + roi_size // 2

    # Crop the center from both images
    original_roi = image_rgb[y1:y2, x1:x2]
    blurred_roi = blurred[y1:y2, x1:x2]

    # Show side-by-side
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(original_roi)
    plt.title("Original ROI", fontsize=14)
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(blurred_roi)
    plt.title("Blurred ROI", fontsize=14)
    plt.axis("off")

    plt.show()
```

```python
#T7


image_path = "document.jpg"

# Load directly as grayscale
gray = cv2.imread(
    image_path,
    cv2.IMREAD_GRAYSCALE
)

if gray is None:
    print("Error: Image not found.")

else:

    # -------------------------
    # GLOBAL THRESHOLD
    # -------------------------

    _, global_threshold = cv2.threshold(
        gray,
        127,
        255,
        cv2.THRESH_BINARY
    )


    # -------------------------
    # ADAPTIVE THRESHOLD
    # -------------------------

    adaptive_threshold = cv2.adaptiveThreshold(
        gray,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        11,
        2
    )


    # -------------------------
    # COMPARE THEM
    # -------------------------

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(global_threshold, cmap="gray")
    plt.title("Global Threshold")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(adaptive_threshold, cmap="gray")
    plt.title("Adaptive Threshold")
    plt.axis("off")

    plt.show()


    # -------------------------
    # ROTATE ADAPTIVE OUTPUT
    # -------------------------

    height, width = adaptive_threshold.shape

    center = (
        width // 2,
        height // 2
    )

    # 45 degree rotation, scale 0.8
    M = cv2.getRotationMatrix2D(
        center,
        45,
        0.8
    )

    rotated = cv2.warpAffine(
        adaptive_threshold,
        M,
        (width, height)
    )


    # Display rotated image
    plt.figure(figsize=(6, 6))
    plt.imshow(rotated, cmap="gray")
    plt.title("Rotated Adaptive Threshold")
    plt.axis("off")
    plt.show()
```

```python
#T8

image1 = cv2.imread(r"/content/Gojo.jpg")
image2 = cv2.imread(r"/content/JJK.jpg")

if image1 is None or image2 is None:
    print("Error: One or both images were not found.")

else:

    # Make both images exactly 500 x 500
    image1 = cv2.resize(image1, (500, 500))
    image2 = cv2.resize(image2, (500, 500))


    # -------------------------
    # CREATE BLACK MASK
    # -------------------------

    mask = np.zeros(
        (500, 500),
        dtype=np.uint8
    )


    # Draw white circle in center
    cv2.circle(
        mask,
        (250, 250),
        150,
        255,
        -1
    )


    # -------------------------
    # CUT CIRCLE FROM IMAGE 1
    # -------------------------

    cutout1 = cv2.bitwise_and(
        image1,
        image1,
        mask=mask
    )


    # -------------------------
    # INVERT MASK
    # -------------------------

    inverse_mask = cv2.bitwise_not(mask)


    # -------------------------
    # GET BACKGROUND FROM IMAGE 2
    # -------------------------

    background2 = cv2.bitwise_and(
        image2,
        image2,
        mask=inverse_mask
    )


    # -------------------------
    # COMBINE THEM
    # -------------------------

    result = cv2.bitwise_or(
        cutout1,
        background2
    )


    # Display
    plt.figure(figsize=(6, 6))
    plt.imshow(
        cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
    )
    plt.title("Bitwise Masking Result")
    plt.axis("off")
    plt.show()
```
