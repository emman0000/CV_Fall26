```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("coins.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray = img
print(img.shape)
plt.axis("off")
plt.imshow(img)
```

```python
threshold_val, binary = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print(threshold_val)
```

```python
plt.hist(gray.ravel(), 256, [0,256],color = 'pink')
plt.title("Grayscale Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()
```

```python
otsu_threshold , otsu_binary = cv2.threshold(gray, 0 , 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print("Otsu threshold:", otsu_threshold)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("OG Image Grayscale")
plt.axis("off")
plt.subplot(1,2,1)
plt.imshow(otsu_binary, cmap='gray')
plt.title("Otsu Image")
plt.axis("off")
plt.show()
```

```python
contrast = cv2.convertScaleAbs(gray, alpha=2.5, beta=0)
plt.imshow(contrast , cmap='gray')
plt.title("Contrast Image")
plt.axis("off")
plt.show()
```

```python
otsu_threshold , otsu_binary2 = cv2.threshold(contrast, 0 , 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
print("Otsu threshold:", otsu_threshold)
```

```python
plt.imshow(otsu_binary, cmap='gray')
plt.title("Otsu Image")
plt.axis("off")
plt.show()
plt.imshow(otsu_binary2, cmap='gray')
plt.title("Otsu Image")
plt.axis("off")
plt.show()
```

Otsu's Thresholding: An automatic global thresholding method that determines the threshold from the image's intensity distribution instead of requiring the user to manually select a threshold. It works particularly well when the histogram has two distinct intensity groups.
Otsu is useful when the correct threshold is unknown because it automatically selects a threshold based on the image histogram, reducing the need for manual trial and error.
